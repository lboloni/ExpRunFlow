# Aggregate the replications

The template of the aggregation stage (docs/DESIGN-Replication.md, Section 7). It collects the `metrics.json` of the replications declared by an aggregate exp/run, and writes the tidy tables and their statistics into its data directory: `replications.csv`, `entities.csv`, `series.csv`, `summary.csv`, `summary-series.csv`, `paired.csv` and `variance.csv`.

Copy it into the project, set the project's import of `Config` (its `exp_run_config`, which sets `Config.PROJECTNAME`), and name it in the `input-to-notebook` of the aggregate family.

```python
import pathlib
import pandas as pd
from exprunflow.exp_run_config import Config  # in a project: from exp_run_config import Config
from exprunflow.aggregate import aggregate_tables, host_summary, load_replications, save_tables
```

```python
creation_style = "exist-ok"
# creation_style = "version"
# creation_style = "discard-old"
expruns_path = None
results_path = None
experiment = "aggregate"
run = "replicated"
```

```python
if expruns_path is not None:
    Config().set_exprun_path(pathlib.Path(expruns_path))
if results_path is not None:
    Config().set_results_path(pathlib.Path(results_path))
exp = Config().get_experiment(experiment, run, creation_style=creation_style)
metrics = load_replications(exp)
tables = aggregate_tables(metrics, exp)
save_tables(tables, exp.data_dir())
print(f"Aggregated {len(metrics)} replications into {exp.data_dir()}")
pd.DataFrame(host_summary(metrics))
```
